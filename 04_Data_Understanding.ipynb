## Column and Row Naming Convention

To keep the notebook professional and readable, column names should follow a consistent **snake_case** convention:

- lowercase letters
- spaces replaced with underscores
- special characters removed
- repeated underscores collapsed
- leading/trailing underscores removed

Row labels should also be explicit. For this project, the DataFrame index is named `row_id` so printed tables clearly distinguish the row label from actual dataset columns.

```python
import re

def clean_column_name(column_name):
    """
    Convert a raw column label to a consistent snake_case name.

    Example:
        "Average Watch Percentage" -> "average_watch_percentage"
    """
    name = str(column_name).strip().lower()
    name = re.sub(r"[^a-z0-9]+", "_", name)
    name = re.sub(r"_+", "_", name)
    return name.strip("_")


def standardize_column_names(data):
    """
    Return a copy of the DataFrame with standardized column names.
    The original DataFrame is not modified.
    """
    result = data.copy()
    result.columns = [clean_column_name(col) for col in result.columns]
    return result


def set_row_index_name(data, index_name="row_id"):
    """
    Return a copy of the DataFrame with a descriptive index name.
    This does not alter the number or order of rows.
    """
    result = data.copy()
    result.index.name = index_name
    return result


# Apply naming standards.
df = standardize_column_names(df)
df = set_row_index_name(df, "row_id")

print("Standardized column names:")
print(df.columns.tolist())

print("\nIndex name:", df.index.name)
display(df.head())
```

## Title-to-Show-ID Validation

A title is not always a unique show identifier.  
The dataset contains titles that correspond to more than one `show_id`, so show-level analysis should group by **both `show_id` and `title`** rather than title alone.

```python
title_show_counts = (
    df.groupby("title")["show_id"]
    .nunique()
    .sort_values(ascending=False)
)

multi_id_titles = title_show_counts[title_show_counts > 1]

print("Titles linked to more than one show_id:", len(multi_id_titles))
display(multi_id_titles)

if len(multi_id_titles) > 0:
    display(
        df[df["title"].isin(multi_id_titles.index)]
        [["show_id", "title", "platform", "release_year"]]
        .drop_duplicates()
        .sort_values(["title", "show_id"])
    )
```
