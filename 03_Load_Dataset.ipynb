```python
# PROCESS: Load the dataset using a portable relative path first.
# The requested Windows project location is also stored below as a convenient copy path.
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 140)

# Copy/paste project folder requested for local Windows use.
PROJECT_COPY_PATH = Path(r"D:\project\_2\_Ott\_Viewer\_Engagement\_Analysis")

# Preferred approach: keep this notebook and CSV in the same folder.
RELATIVE_DATA_PATH = Path("ott_viewer_dropoff_retention_us_v1.0.csv")

# Optional absolute path if you run the notebook from another working directory.
WINDOWS_DATA_PATH = PROJECT_COPY_PATH / "ott_viewer_dropoff_retention_us_v1.0.csv"

if RELATIVE_DATA_PATH.exists():
    DATA_PATH = RELATIVE_DATA_PATH
elif WINDOWS_DATA_PATH.exists():
    DATA_PATH = WINDOWS_DATA_PATH
else:
    raise FileNotFoundError(
        "CSV not found. Keep it beside the notebook or copy the project to: "
        + str(PROJECT_COPY_PATH)
    )

df = pd.read_csv(DATA_PATH)

print("Dataset loaded from:", DATA_PATH)
print("Dataset shape:", df.shape)
df.head()
```

```python
# KERNEL-SAFE SETUP
# Keep plotting memory controlled while preserving the analysis.
import gc

MAX_PLOT_ROWS = 5000

def safe_sample(data, n=MAX_PLOT_ROWS, random_state=42):
    """Use the full data when small; otherwise use a reproducible sample for plotting."""
    return data if len(data) <= n else data.sample(n=n, random_state=random_state)

def finish_plot():
    """Render and release the current Matplotlib figure."""
    plt.tight_layout()
    plt.show()
    plt.close()
    gc.collect()
```
