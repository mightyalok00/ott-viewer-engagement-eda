## Executive Summary

This portfolio project analyzes **33,171 synthetic OTT episode-level records** to understand engagement, watch completion, drop-off, and retention risk in a Hotstar business scenario.

### Headline Findings

- **Comedy** has the highest average engagement score at approximately **77.38**.
- **Drama** has the highest observed drop-off rate at approximately **24.19%**.
- Approximately **14.48%** of episode-level records are classified as **High retention risk**.
- Watch completion and drop-off probability have a very strong negative correlation of approximately **-0.957**, meaning lower completion is strongly associated with higher drop-off probability in this synthetic dataset.
- Hook strength has a moderate positive association with watch percentage (**r ≈ 0.552**).
- Cognitive load has a strong positive association with drop-off probability (**r ≈ 0.793**).

The business goal is to identify content worth promoting and episodes that require investigation, while treating these EDA patterns as **associations rather than causal proof**.
